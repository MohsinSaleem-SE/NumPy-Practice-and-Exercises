MISSING VALUES

```python
import numpy as np
```

```python
a = np.array([1,2,np.nan,4,np.nan,6])
a
```

```python
np.isnan(a)
```

```python
np.isnan(a).sum()
```

```python
b = np.nanmean(a)
b
```

```python
np.nansum(a)
```

```python
np.nan_to_num(a, nan = 9)
```

```python
a[np.isnan(a)] = b
a
```

```python
np.nan_to_num(a, nan = b)
```
